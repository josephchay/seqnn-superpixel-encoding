```python
from seqnn_pytorch import SEQNN, SEQNNTrainer, SEQNNConfig
from seqnn_dataLoader import DataLoader

# Load data
dataloader = DataLoader('sat')
train_x, train_y, valid_x, valid_y, test_x, test_y = dataloader.get_data()

# Build model
config = SEQNNConfig()
model = SEQNN(n_classes=6, n_channels=4, config=config, use_quantum=False)

# Train
trainer = SEQNNTrainer(model)
trainer.compile(learning_rate=0.01)
trainer.fit(train_x, train_y, valid_x, valid_y, epochs=200, batch_size=50)
```
